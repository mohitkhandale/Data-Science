```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Feature1' : [1.0,2.0,np.nan,4.0],
    'Feature2' : [np.nan,2.0,3.0,4.0],
    'Feature3' : [1.0,2.0,3.0, np.nan]
}
dataframe = pd.DataFrame(data)
dataframe
```

```python
from sklearn.impute import KNNImputer
impute = KNNImputer()
dataframe_imputed = impute.fit_transform(dataframe)
dataframe_imputed
```

fit -> evaluations based on computing the nearest neighbor.

transform -> acutally fill(impute) the missing values in your data
