```python
X = data.drop(columns='Outcome', axis=1)
Y = data['Outcome']
X
Y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, random_state=44)
```

```python
# Data Imbalance
data['Outcome'].value_counts() 
```

1. Oversampling
2. Undersampling

```python
from collections import Counter
Counter(Y_train)
```

RandomOverSampler: Randomly Dupliacting Samples from the minority class until the both the classes have the same number of samples

```python
from imblearn.over_sampling import RandomOverSampler
from imblearn.under_sampling import RandomUnderSampler

model_over_sampling = RandomOverSampler(random_state=44)
X_train_over, Y_train_over = model_over_sampling.fit_resample(X_train, Y_train)
```

```python
Counter(Y_train_over)
```

Advanced Over Sampling Techniques: SMOTE & ADASYN (Synthetic Data)

Model Training

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X_train_over, Y_train_over)
```

Model Prediction

```python
y_pred = model.predict(X_test)
y_pred
```

```python
result_comparison =  pd.DataFrame({'Actual Outcome': Y_test, 'Predicted Outcome': y_pred})
result_comparison
```

```python
y_pred_prob = model.predict_proba(X_test)
y_pred_prob
```

Model Evaluation

```python
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score
confusion_matrix(Y_test, y_pred)
```

```python
recall_score(Y_test, y_pred)
```
