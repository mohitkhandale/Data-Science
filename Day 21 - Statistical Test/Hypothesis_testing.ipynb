```python
from scipy.stats import norm

population_mean = 800
sample_mean = 790
standard_error = 7.5
alpha = 0.05

z_score = (sample_mean - population_mean)/standard_error

# Two tailed teest
p_value = 2 * norm.cdf(z_score)

if p_value > alpha:
    print('Fail to reject the null hypothesis: Average lifespam of the light bulbs is approximately 800 hours')
else:
    print('Reject thr null hypothesis: Average lifespam of the light is not 800 hours')
```
